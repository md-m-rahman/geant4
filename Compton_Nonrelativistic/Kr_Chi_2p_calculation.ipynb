```python
import numpy as np
from scipy.special import spherical_jn
from scipy.integrate import quad, cumulative_trapezoid
import pandas as pd

# Kr R_2p(r) terms: C * r^n * exp(-Z r)
terms_2p = [
    (557.4614,   1, 38.0283),
    (718.5621,   1, 14.6932),
    (8576.2463,  2, 31.3136),
    (105.3739,   2, 11.7833),
    (0.6261,     2, 7.4972),
    (0.1229,     3, 5.3888),
    (18056.1140, 3, 25.3638),
    (-0.0015,    3, 3.5216),
    (-0.0001,    3, 2.2424),
    (-0.0000,    3, 1.4475),
]

def R_2p(r):
    total = 0.0
    for C, n, Z in terms_2p:
        total += C * r**n * np.exp(-Z*r)
    return total

def chi_2p(p):
    integrand = lambda r: R_2p(r) * spherical_jn(1, p*r) * r**2
    val, err = quad(integrand, 0, np.inf, limit=300)
    return np.sqrt(2/np.pi) * val

# momentum grid, step = 0.05
p_grid = np.arange(0.0, 100.0 + 0.05, 0.05)

# Chi_2p(p)
chi = np.array([chi_2p(p) for p in p_grid])

# I_2p(p) = chi^2 p^2
I2p = chi**2 * p_grid**2

# J_2p(Q) = 1/2 ∫_Q^∞ chi^2(p) p dp
J_integrand = chi**2 * p_grid

J2p = -0.5 * cumulative_trapezoid(
    J_integrand[::-1],
    p_grid[::-1],
    initial=0
)[::-1]

# make table
result = pd.DataFrame({
    "p_or_Q": p_grid,
    "chi_2p(p)": chi,
    "I_2p(p)": I2p,
    "J_2p(Q)": J2p
})

# print all rows
pd.set_option("display.max_rows", None)
print(result.to_string(index=False))

# save to CSV
result.to_csv("Kr_2p_chi_I_J.csv", index=False)

print("\nSaved file: Kr_2p_chi_I_J.csv")
```
