```python
import os
import ROOT
import pandas as pd

# =====================================================
# WORKING DIRECTORY
# =====================================================
working_directory = "/root/geant4/detector/Lung_ICRP"
os.chdir(working_directory)

print("Current working directory:")
print(os.getcwd())

# =====================================================
# USER SETTINGS
# =====================================================
file_path = "lung1_tissueTumor.root"
tree_name = "t"

selected_volumes = [2, 3, 4, 5]
selected_pdg = 22          # gamma

excel_file = "lung1_tissueTumor_all_gamma_data.xlsx"

# =====================================================
# CHECK INPUT FILE
# =====================================================
if not os.path.exists(file_path):
    raise FileNotFoundError(
        f"ROOT file not found:\n{os.path.abspath(file_path)}"
    )

# =====================================================
# OPEN ROOT FILE
# =====================================================
root_file = ROOT.TFile.Open(file_path)

if not root_file or root_file.IsZombie():
    raise OSError(
        f"Could not open ROOT file:\n{os.path.abspath(file_path)}"
    )

tree = root_file.Get(tree_name)

if not tree:
    root_file.Close()
    raise KeyError(f"TTree '{tree_name}' was not found.")

print("\nInput ROOT file:")
print(os.path.abspath(file_path))

print("\nTotal TTree entries:")
print(tree.GetEntries())

# =====================================================
# CHECK BRANCHES
# =====================================================
available_branches = {
    branch.GetName()
    for branch in tree.GetListOfBranches()
}

required_branches = [
    "vlm",
    "pdg",
    "pro",
    "stp",
    "k",
    "et",
    "de",
    "x",
    "y",
    "z",
    "px",
    "py",
    "pz"
]

missing_branches = [
    branch for branch in required_branches
    if branch not in available_branches
]

if missing_branches:
    root_file.Close()
    raise KeyError(
        f"Missing branches: {missing_branches}\n"
        f"Available branches: {sorted(available_branches)}"
    )

# =====================================================
# OUTPUT COLUMN ORDER
# =====================================================
column_order = [
    "tree_entry",
    "vector_index",
    "vlm",
    "pdg",
    "pro",
    "stp",
    "k",
    "et",
    "de",
    "x",
    "y",
    "z",
    "px",
    "py",
    "pz"
]

# =====================================================
# STORAGE
# =====================================================
volume_data = {
    volume: []
    for volume in selected_volumes
}

combined_data = []

# =====================================================
# LOOP THROUGH TTREE ENTRIES
# =====================================================
for tree_entry, event in enumerate(tree):

    # Number of tracking records inside this TTree entry
    number_of_records = len(event.pdg)

    # Check that the important vectors have matching lengths
    vector_lengths = {
        "pdg": len(event.pdg),
        "vlm": len(event.vlm),
        "pro": len(event.pro),
        "stp": len(event.stp),
        "k": len(event.k),
        "et": len(event.et),
        "de": len(event.de),
        "x": len(event.x),
        "y": len(event.y),
        "z": len(event.z),
        "px": len(event.px),
        "py": len(event.py),
        "pz": len(event.pz)
    }

    if len(set(vector_lengths.values())) != 1:
        print(
            f"Warning: vector lengths differ in TTree entry "
            f"{tree_entry}: {vector_lengths}"
        )

        # Use the shortest vector length to avoid an index error
        number_of_records = min(vector_lengths.values())

    # Loop through all records in this event
    for vector_index in range(number_of_records):

        pdg_value = int(event.pdg[vector_index])
        volume_value = int(event.vlm[vector_index])

        # Keep only gamma records
        if pdg_value != selected_pdg:
            continue

        # Keep only volumes 2, 3, 4 and 5
        if volume_value not in selected_volumes:
            continue

        row = {
            "tree_entry": tree_entry,
            "vector_index": vector_index,

            "vlm": volume_value,
            "pdg": pdg_value,
            "pro": int(event.pro[vector_index]),
            "stp": int(event.stp[vector_index]),

            "k": float(event.k[vector_index]),
            "et": float(event.et[vector_index]),
            "de": float(event.de[vector_index]),

            "x": float(event.x[vector_index]),
            "y": float(event.y[vector_index]),
            "z": float(event.z[vector_index]),

            "px": float(event.px[vector_index]),
            "py": float(event.py[vector_index]),
            "pz": float(event.pz[vector_index])
        }

        volume_data[volume_value].append(row)
        combined_data.append(row)

# =====================================================
# CREATE DATAFRAMES
# =====================================================
dataframes = {}

for volume in selected_volumes:
    dataframes[volume] = pd.DataFrame(
        volume_data[volume],
        columns=column_order
    )

df_all = pd.DataFrame(
    combined_data,
    columns=column_order
)

# =====================================================
# SAVE EXCEL WORKBOOK
# =====================================================
with pd.ExcelWriter(
    excel_file,
    engine="openpyxl"
) as writer:

    for volume in selected_volumes:

        dataframes[volume].to_excel(
            writer,
            sheet_name=f"Volume_{volume}",
            index=False
        )

    df_all.to_excel(
        writer,
        sheet_name="All_Volumes",
        index=False
    )

# =====================================================
# CLOSE ROOT FILE
# =====================================================
root_file.Close()

# =====================================================
# PRINT SUMMARY
# =====================================================
print("\nRecords saved:")

for volume in selected_volumes:
    print(
        f"Volume {volume}: "
        f"{len(dataframes[volume])} gamma records"
    )

print(
    f"All selected volumes: "
    f"{len(df_all)} gamma records"
)

print("\nExcel workbook saved as:")
print(os.path.abspath(excel_file))

# Display first 10 rows
display(df_all.head(10))
```
