```python
import ROOT
import math
import os
from IPython.display import Image, display

%jsroot on

# ============================================================
# USER SETTINGS
# ============================================================

file_path = "/root/geant4/detector/Tumor/Tumor1/tumor1.root"
tree_name = "t"

selected_volume = 2
selected_process = 2013       # Compton scattering
#selected_step = 1
selected_pdg = 22             # gamma

# Incident beam direction from your GPS:
# /gps/direction 0 -1 0
incident_direction = (0.0, -1.0, 0.0)

output_file = "de_et_vs_theta_vlm2.png"


# ============================================================
# OPEN ROOT FILE
# ============================================================

if not os.path.exists(file_path):
    raise FileNotFoundError(
        f"ROOT file was not found:\n{file_path}"
    )

f = ROOT.TFile.Open(file_path)

if not f or f.IsZombie():
    raise RuntimeError(
        f"ROOT could not open the file:\n{file_path}"
    )

t = f.Get(tree_name)

if not t:
    f.ls()
    raise RuntimeError(
        f"Tree '{tree_name}' was not found in the ROOT file."
    )

print("File opened successfully:", file_path)
print("Number of events:", t.GetEntries())


# ============================================================
# CHECK REQUIRED BRANCHES
# ============================================================

required_branches = [
    "pdg", "pro", "vlm", "stp",
    "de", "et", "px", "py", "pz"
]

available_branches = {
    branch.GetName() for branch in t.GetListOfBranches()
}

missing_branches = [
    name for name in required_branches
    if name not in available_branches
]

if missing_branches:
    raise RuntimeError(
        "The following required branches are missing: "
        + ", ".join(missing_branches)
    )

print("All required branches are available.")


# ============================================================
# DELETE OLD ROOT OBJECTS
# ============================================================

for object_name in [
    "h_de_theta",
    "h_et_theta",
    "c_de_et_theta"
]:
    old_object = ROOT.gROOT.FindObject(object_name)

    if old_object:
        old_object.Delete()


# ============================================================
# CREATE 2D HISTOGRAMS
# ============================================================

h_de_theta = ROOT.TH2F(
    "h_de_theta",
    "Deposited energy versus Compton angle, volume 2;"
    "Compton scattering angle #theta (degrees);"
    "Deposited energy de (keV)",
    90, 0.0, 180.0,
    140, 0.0, 700.0
)

h_et_theta = ROOT.TH2F(
    "h_et_theta",
    "Photon energy versus Compton angle, volume 2;"
    "Compton scattering angle #theta (degrees);"
    "Photon energy et (keV)",
    90, 0.0, 180.0,
    140, 0.0, 700.0
)

h_de_theta.SetStats(0)
h_et_theta.SetStats(0)


# ============================================================
# NORMALIZE INCIDENT DIRECTION
# ============================================================

ix, iy, iz = incident_direction

incident_magnitude = math.sqrt(
    ix**2 + iy**2 + iz**2
)

if incident_magnitude == 0:
    raise ValueError("Incident direction cannot be zero.")

ix /= incident_magnitude
iy /= incident_magnitude
iz /= incident_magnitude


# ============================================================
# LOOP OVER EVENTS
# ============================================================

selected_points = 0
skipped_zero_momentum = 0
skipped_invalid_angle = 0

for event_number, event in enumerate(t):

    # All branches should have corresponding entries.
    # min() prevents an index error if one branch is shorter.
    n_steps = min(
        len(event.pdg),
        len(event.pro),
        len(event.vlm),
        len(event.stp),
        len(event.de),
        len(event.et),
        len(event.px),
        len(event.py),
        len(event.pz)
    )

    for i in range(n_steps):

        # ----------------------------------------------------
        # Selection
        # ----------------------------------------------------

        if int(event.pdg[i]) != selected_pdg:
            continue

        if int(event.pro[i]) != selected_process:
            continue

        if int(event.vlm[i]) != selected_volume:
            continue

        if int(event.stp[i]) != selected_step:
            continue

        # ----------------------------------------------------
        # Scattered photon momentum direction
        # ----------------------------------------------------

        px = float(event.px[i])
        py = float(event.py[i])
        pz = float(event.pz[i])

        momentum_magnitude = math.sqrt(
            px**2 + py**2 + pz**2
        )

        if momentum_magnitude <= 0:
            skipped_zero_momentum += 1
            continue

        ux = px / momentum_magnitude
        uy = py / momentum_magnitude
        uz = pz / momentum_magnitude

        # Dot product between incident and scattered directions
        cos_theta = (
            ix * ux +
            iy * uy +
            iz * uz
        )

        # Protect against numerical values such as 1.00000001
        cos_theta = max(-1.0, min(1.0, cos_theta))

        theta_deg = math.degrees(
            math.acos(cos_theta)
        )

        if not math.isfinite(theta_deg):
            skipped_invalid_angle += 1
            continue

        de_value = float(event.de[i])
        et_value = float(event.et[i])

        # ----------------------------------------------------
        # Fill histograms
        # ----------------------------------------------------

        if math.isfinite(de_value):
            h_de_theta.Fill(theta_deg, de_value)

        if math.isfinite(et_value):
            h_et_theta.Fill(theta_deg, et_value)

        selected_points += 1


# ============================================================
# PRINT RESULTS
# ============================================================

print("\nSelection used:")
print("pdg =", selected_pdg)
print("pro =", selected_process)
print("vlm =", selected_volume)
print("stp =", selected_step)

print("\nSelected points:", selected_points)
print("de histogram entries:", h_de_theta.GetEntries())
print("et histogram entries:", h_et_theta.GetEntries())
print("Skipped zero-momentum points:", skipped_zero_momentum)
print("Skipped invalid angles:", skipped_invalid_angle)


# ============================================================
# DRAW HISTOGRAMS
# ============================================================

c = ROOT.TCanvas(
    "c_de_et_theta",
    "de and et versus Compton scattering angle",
    1200,
    1000
)

c.Divide(1, 2)


# ------------------------------------------------------------
# Top panel: de versus theta
# ------------------------------------------------------------

c.cd(1)

ROOT.gPad.SetLeftMargin(0.11)
ROOT.gPad.SetRightMargin(0.15)
ROOT.gPad.SetBottomMargin(0.12)
ROOT.gPad.SetTopMargin(0.09)

if h_de_theta.GetEntries() > 0:
    ROOT.gPad.SetLogz()

h_de_theta.Draw("COLZ")


# ------------------------------------------------------------
# Bottom panel: et versus theta
# ------------------------------------------------------------

c.cd(2)

ROOT.gPad.SetLeftMargin(0.11)
ROOT.gPad.SetRightMargin(0.15)
ROOT.gPad.SetBottomMargin(0.12)
ROOT.gPad.SetTopMargin(0.09)

if h_et_theta.GetEntries() > 0:
    ROOT.gPad.SetLogz()

h_et_theta.Draw("COLZ")


# ============================================================
# SAVE AND DISPLAY
# ============================================================

c.Modified()
c.Update()

c.SaveAs(output_file)
```
