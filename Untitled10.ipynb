```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from scipy.special import spherical_jn
from scipy.integrate import cumulative_trapezoid


# =====================================================
# INPUT FILE
# =====================================================

csv_file = "/root/geant4/fac/No_all_wavefunctions.csv"

df = pd.read_csv(csv_file)
df = df.dropna(subset=["r", "P", "Q"])

print("Available orbitals:")
print(df["Orbital"].unique())


# =====================================================
# SETTINGS
# =====================================================

dp = 0.05

# Final Compton-profile range
Qgrid = np.arange(0.0, 100.0 + dp, dp)

# Number of points used for interpolated radial grid
N_R = 10000


# =====================================================
# READ AND NORMALIZE ONE FAC ORBITAL
# =====================================================

def get_orbital(df, name, print_values=True):

    d = df[df["Orbital"] == name].copy()
    d = d.sort_values("r")

    if len(d) == 0:
        raise ValueError(f"Orbital {name} not found in CSV file.")

    r_raw = d["r"].to_numpy(dtype=float)
    P_raw = d["P"].to_numpy(dtype=float)
    Q_raw = d["Q"].to_numpy(dtype=float)

    if print_values:

        print("\n===================================")
        print(name)
        print("===================================")

        print("\nFirst 8 points:")
        print(d[["r", "P", "Q"]].head(8))

        print("\nLast 8 points:")
        print(d[["r", "P", "Q"]].tail(8))


    # -------------------------------------------------
    # FAC radial normalization
    #
    # ∫ [P(r)^2 + Q(r)^2] dr = 1
    # -------------------------------------------------

    raw_norm = np.trapezoid(
        P_raw**2 + Q_raw**2,
        r_raw
    )

    print(
        name,
        "raw radial norm ∫(P²+Q²)dr =",
        raw_norm
    )


    # Normalize raw FAC functions
    P_raw = P_raw / np.sqrt(raw_norm)
    Q_raw = Q_raw / np.sqrt(raw_norm)


    # -------------------------------------------------
    # Interpolate onto finer uniform radial grid
    # -------------------------------------------------

    r = np.linspace(
        r_raw.min(),
        r_raw.max(),
        N_R
    )

    P = np.interp(
        r,
        r_raw,
        P_raw
    )

    Qs = np.interp(
        r,
        r_raw,
        Q_raw
    )


    # Normalize again after interpolation
    fine_norm = np.trapezoid(
        P**2 + Qs**2,
        r
    )

    P = P / np.sqrt(fine_norm)
    Qs = Qs / np.sqrt(fine_norm)

    final_norm = np.trapezoid(
        P**2 + Qs**2,
        r
    )

    print(
        name,
        "fine radial norm after interpolation =",
        final_norm
    )


    # -------------------------------------------------
    # Useful numerical information
    # -------------------------------------------------

    dr = r[1] - r[0]

    print(
        name,
        "rmax =",
        r.max()
    )

    print(
        name,
        "dr =",
        dr
    )

    print(
        name,
        "approx pi/dr =",
        np.pi / dr
    )

    return r, P, Qs


# =====================================================
# FOURIER-BESSEL TRANSFORM
# =====================================================

def chi_transform(p_grid, r, radial, l):

    chi = np.zeros(len(p_grid))

    prefactor = np.sqrt(2.0 / np.pi)

    for i, pp in enumerate(p_grid):

        jl = spherical_jn(
            l,
            pp * r
        )

        # FAC:
        #
        # P(r) = r G(r)
        # Q(r) = r F(r)
        #
        # χ_l(p)
        # =
        # sqrt(2/pi)
        # ∫ P(r) j_l(pr) r dr
        #

        chi[i] = prefactor * np.trapezoid(
            radial * jl * r,
            r
        )

    return chi


# =====================================================
# CALCULATE COMPTON PROFILE
# =====================================================

def calculate_J(
    p_grid,
    chiG,
    chiF,
    Qgrid
):

    # I(p) = p² [χG² + χF²]

    # Therefore:
    #
    # J(Q)
    # =
    # 1/2 ∫_Q^∞ I(p)/p dp
    #
    # =
    # 1/2 ∫_Q^∞
    # p [χG² + χF²] dp

    integrand = (
        chiG**2 + chiF**2
    ) * p_grid


    rev = cumulative_trapezoid(
        integrand[::-1],
        p_grid[::-1],
        initial=0
    )

    J_p = -0.5 * rev[::-1]


    # Interpolate J(p) onto requested Q grid
    J_Q = np.interp(
        Qgrid,
        p_grid,
        J_p
    )

    return J_Q


# =====================================================
# CALCULATE ONE ORBITAL
# =====================================================

def one_orbital_J(
    df,
    name,
    l_large,
    l_small,
    pmax
):

    print("\n")
    print("###################################")
    print("Calculating", name)
    print("pmax =", pmax, "a.u.")
    print("###################################")


    # Each orbital gets its own momentum grid
    p_local = np.arange(
        0.0,
        pmax + dp,
        dp
    )


    # Get radial functions
    r, P, Qs = get_orbital(
        df,
        name
    )


    # Large component
    chiG = chi_transform(
        p_local,
        r,
        P,
        l_large
    )


    # Small component
    chiF = chi_transform(
        p_local,
        r,
        Qs,
        l_small
    )


    # -------------------------------------------------
    # Momentum density
    # -------------------------------------------------

    I = (
        chiG**2 +
        chiF**2
    ) * p_local**2


    momentum_norm = np.trapezoid(
        I,
        p_local
    )


    print()
    print(
        name,
        "momentum norm ∫I(p)dp =",
        momentum_norm
    )


    # Useful warning
    if momentum_norm < 0.995:

        print(
            "WARNING:",
            name,
            "momentum norm is LOW."
        )

        print(
            "Consider increasing pmax."
        )


    elif momentum_norm > 1.005:

        print(
            "WARNING:",
            name,
            "momentum norm is HIGH."
        )

        print(
            "Possible high-p numerical aliasing."
        )

        print(
            "Reduce pmax or increase radial resolution."
        )


    else:

        print(
            name,
            "momentum normalization is GOOD."
        )


    # -------------------------------------------------
    # Compton profile
    # -------------------------------------------------

    J = calculate_J(
        p_local,
        chiG,
        chiF,
        Qgrid
    )


    return J, momentum_norm


# =====================================================
# RELATIVISTIC ORBITALS
#
# Large-component l:
#
# s -> 0
# p -> 1
# d -> 2
# f -> 3
#
# Small component:
#
# s1/2 -> p      l'=1
#
# p1/2 -> s      l'=0
# p3/2 -> d      l'=2
#
# d3/2 -> p      l'=1
# d5/2 -> f      l'=3
#
# f5/2 -> d      l'=2
# f7/2 -> g      l'=4
# =====================================================


# =====================================================
# 1s
# =====================================================

J_1s12, N_1s12 = one_orbital_J(
    df,
    "No_1s12",
    0,
    1,
    pmax=1000
)


# =====================================================
# 2s, 2p
# =====================================================

J_2s12, N_2s12 = one_orbital_J(
    df,
    "No_2s12",
    0,
    1,
    pmax=800
)

J_2p12, N_2p12 = one_orbital_J(
    df,
    "No_2p12",
    1,
    0,
    pmax=800
)

J_2p32, N_2p32 = one_orbital_J(
    df,
    "No_2p32",
    1,
    2,
    pmax=800
)


# =====================================================
# 3s, 3p, 3d
# =====================================================

J_3s12, N_3s12 = one_orbital_J(
    df,
    "No_3s12",
    0,
    1,
    pmax=600
)

J_3p12, N_3p12 = one_orbital_J(
    df,
    "No_3p12",
    1,
    0,
    pmax=600
)

J_3p32, N_3p32 = one_orbital_J(
    df,
    "No_3p32",
    1,
    2,
    pmax=600
)

J_3d32, N_3d32 = one_orbital_J(
    df,
    "No_3d32",
    2,
    1,
    pmax=600
)

J_3d52, N_3d52 = one_orbital_J(
    df,
    "No_3d52",
    2,
    3,
    pmax=600
)


# =====================================================
# 4s, 4p, 4d, 4f
# =====================================================

J_4s12, N_4s12 = one_orbital_J(
    df,
    "No_4s12",
    0,
    1,
    pmax=400
)

J_4p12, N_4p12 = one_orbital_J(
    df,
    "No_4p12",
    1,
    0,
    pmax=400
)

J_4p32, N_4p32 = one_orbital_J(
    df,
    "No_4p32",
    1,
    2,
    pmax=400
)

J_4d32, N_4d32 = one_orbital_J(
    df,
    "No_4d32",
    2,
    1,
    pmax=400
)

J_4d52, N_4d52 = one_orbital_J(
    df,
    "No_4d52",
    2,
    3,
    pmax=400
)

J_4f52, N_4f52 = one_orbital_J(
    df,
    "No_4f52",
    3,
    2,
    pmax=300
)

J_4f72, N_4f72 = one_orbital_J(
    df,
    "No_4f72",
    3,
    4,
    pmax=300
)


# =====================================================
# 5s, 5p, 5d, 5f
# =====================================================

J_5s12, N_5s12 = one_orbital_J(
    df,
    "No_5s12",
    0,
    1,
    pmax=300
)

J_5p12, N_5p12 = one_orbital_J(
    df,
    "No_5p12",
    1,
    0,
    pmax=300
)

J_5p32, N_5p32 = one_orbital_J(
    df,
    "No_5p32",
    1,
    2,
    pmax=300
)

J_5d32, N_5d32 = one_orbital_J(
    df,
    "No_5d32",
    2,
    1,
    pmax=300
)

J_5d52, N_5d52 = one_orbital_J(
    df,
    "No_5d52",
    2,
    3,
    pmax=300
)

J_5f52, N_5f52 = one_orbital_J(
    df,
    "No_5f52",
    3,
    2,
    pmax=300
)

J_5f72, N_5f72 = one_orbital_J(
    df,
    "No_5f72",
    3,
    4,
    pmax=300
)


# =====================================================
# 6s, 6p
# =====================================================

J_6s12, N_6s12 = one_orbital_J(
    df,
    "No_6s12",
    0,
    1,
    pmax=300
)

J_6p12, N_6p12 = one_orbital_J(
    df,
    "No_6p12",
    1,
    0,
    pmax=300
)

J_6p32, N_6p32 = one_orbital_J(
    df,
    "No_6p32",
    1,
    2,
    pmax=300
)


# =====================================================
# 7s
# =====================================================

J_7s12, N_7s12 = one_orbital_J(
    df,
    "No_7s12",
    0,
    1,
    pmax=300
)


# =====================================================
# MOMENTUM NORMALIZATION SUMMARY
# =====================================================

norm_table = pd.DataFrame({

    "Orbital": [
        "1s1/2",
        "2s1/2",
        "2p1/2",
        "2p3/2",

        "3s1/2",
        "3p1/2",
        "3p3/2",
        "3d3/2",
        "3d5/2",

        "4s1/2",
        "4p1/2",
        "4p3/2",
        "4d3/2",
        "4d5/2",
        "4f5/2",
        "4f7/2",

        "5s1/2",
        "5p1/2",
        "5p3/2",
        "5d3/2",
        "5d5/2",
        "5f5/2",
        "5f7/2",

        "6s1/2",
        "6p1/2",
        "6p3/2",

        "7s1/2"
    ],

    "Momentum_norm": [
        N_1s12,
        N_2s12,
        N_2p12,
        N_2p32,

        N_3s12,
        N_3p12,
        N_3p32,
        N_3d32,
        N_3d52,

        N_4s12,
        N_4p12,
        N_4p32,
        N_4d32,
        N_4d52,
        N_4f52,
        N_4f72,

        N_5s12,
        N_5p12,
        N_5p32,
        N_5d32,
        N_5d52,
        N_5f52,
        N_5f72,

        N_6s12,
        N_6p12,
        N_6p32,

        N_7s12
    ]
})


print("\n")
print("========================================")
print("MOMENTUM NORMALIZATION SUMMARY")
print("Target for every orbital: approximately 1")
print("========================================")

print(norm_table.to_string(index=False))


norm_table.to_csv(
    "No_momentum_normalization_check.csv",
    index=False
)


# =====================================================
# TOTAL ELECTRON-NUMBER CHECK FROM I(p)
# =====================================================

N_total = (

    2*N_1s12

    + 2*N_2s12
    + 2*N_2p12
    + 4*N_2p32

    + 2*N_3s12
    + 2*N_3p12
    + 4*N_3p32
    + 4*N_3d32
    + 6*N_3d52

    + 2*N_4s12
    + 2*N_4p12
    + 4*N_4p32
    + 4*N_4d32
    + 6*N_4d52
    + 6*N_4f52
    + 8*N_4f72

    + 2*N_5s12
    + 2*N_5p12
    + 4*N_5p32
    + 4*N_5d32
    + 6*N_5d52
    + 6*N_5f52
    + 8*N_5f72

    + 2*N_6s12
    + 2*N_6p12
    + 4*N_6p32

    + 2*N_7s12
)


print("\n========================================")
print("TOTAL MOMENTUM NORMALIZATION")
print("========================================")

print(
    "Occupancy-weighted Σ ∫I(p)dp =",
    N_total
)

print(
    "Expected for neutral Nobelium = 102"
)


# =====================================================
# TOTAL RELATIVISTIC COMPTON PROFILE
# =====================================================

J_total_No_rel = (

    2*J_1s12

    + 2*J_2s12
    + 2*J_2p12
    + 4*J_2p32

    + 2*J_3s12
    + 2*J_3p12
    + 4*J_3p32
    + 4*J_3d32
    + 6*J_3d52

    + 2*J_4s12
    + 2*J_4p12
    + 4*J_4p32
    + 4*J_4d32
    + 6*J_4d52
    + 6*J_4f52
    + 8*J_4f72

    + 2*J_5s12
    + 2*J_5p12
    + 4*J_5p32
    + 4*J_5d32
    + 6*J_5d52
    + 6*J_5f52
    + 8*J_5f72

    + 2*J_6s12
    + 2*J_6p12
    + 4*J_6p32

    + 2*J_7s12
)


# =====================================================
# SAVE COMPTON PROFILE
# =====================================================

table = pd.DataFrame({

    "Q": Qgrid,

    "J_1s1/2": J_1s12,

    "J_2s1/2": J_2s12,
    "J_2p1/2": J_2p12,
    "J_2p3/2": J_2p32,

    "J_3s1/2": J_3s12,
    "J_3p1/2": J_3p12,
    "J_3p3/2": J_3p32,
    "J_3d3/2": J_3d32,
    "J_3d5/2": J_3d52,

    "J_4s1/2": J_4s12,
    "J_4p1/2": J_4p12,
    "J_4p3/2": J_4p32,
    "J_4d3/2": J_4d32,
    "J_4d5/2": J_4d52,
    "J_4f5/2": J_4f52,
    "J_4f7/2": J_4f72,

    "J_5s1/2": J_5s12,
    "J_5p1/2": J_5p12,
    "J_5p3/2": J_5p32,
    "J_5d3/2": J_5d32,
    "J_5d5/2": J_5d52,
    "J_5f5/2": J_5f52,
    "J_5f7/2": J_5f72,

    "J_6s1/2": J_6s12,
    "J_6p1/2": J_6p12,
    "J_6p3/2": J_6p32,

    "J_7s1/2": J_7s12,

    "J_total_No_relativistic":
        J_total_No_rel
})


output_file = "No_relativistic_Compton_profile.csv"

table.to_csv(
    output_file,
    index=False
)

print("\nSaved:")
print(output_file)


# =====================================================
# PLOT
# =====================================================

plt.figure(figsize=(9, 7))


plt.plot(
    Qgrid,
    2*J_1s12,
    label="1s(2)"
)

plt.plot(
    Qgrid,
    2*J_2s12,
    label="2s(2)"
)

plt.plot(
    Qgrid,
    2*J_2p12 + 4*J_2p32,
    label="2p(6)"
)

plt.plot(
    Qgrid,
    2*J_3s12,
    label="3s(2)"
)

plt.plot(
    Qgrid,
    2*J_3p12 + 4*J_3p32,
    label="3p(6)"
)

plt.plot(
    Qgrid,
    4*J_3d32 + 6*J_3d52,
    label="3d(10)"
)

plt.plot(
    Qgrid,
    2*J_4s12,
    label="4s(2)"
)

plt.plot(
    Qgrid,
    2*J_4p12 + 4*J_4p32,
    label="4p(6)"
)

plt.plot(
    Qgrid,
    4*J_4d32 + 6*J_4d52,
    label="4d(10)"
)

plt.plot(
    Qgrid,
    6*J_4f52 + 8*J_4f72,
    label="4f(14)"
)

plt.plot(
    Qgrid,
    2*J_5s12,
    label="5s(2)"
)

plt.plot(
    Qgrid,
    2*J_5p12 + 4*J_5p32,
    label="5p(6)"
)

plt.plot(
    Qgrid,
    4*J_5d32 + 6*J_5d52,
    label="5d(10)"
)

plt.plot(
    Qgrid,
    6*J_5f52 + 8*J_5f72,
    label="5f(14)"
)

plt.plot(
    Qgrid,
    2*J_6s12,
    label="6s(2)"
)

plt.plot(
    Qgrid,
    2*J_6p12 + 4*J_6p32,
    label="6p(6)"
)

plt.plot(
    Qgrid,
    2*J_7s12,
    label="7s(2)"
)


plt.plot(
    Qgrid,
    J_total_No_rel,
    label="No total relativistic",
    linewidth=2
)


plt.xlabel("Q (a.u.)")
plt.ylabel("J(Q)")

plt.title(
    "Nobelium Relativistic Compton Profile"
)

plt.yscale("log")

plt.grid(True)

plt.legend(
    fontsize=8,
    ncol=2
)

plt.tight_layout()

plt.show()
```
