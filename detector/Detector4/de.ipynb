```python
import ROOT
%jsroot on

f = ROOT.TFile.Open("tumor.root")
t = f.Get("t")

# Define your custom names here
vol_names = {
    1: "room",
    2: "Phantom",
    3: "Tumor",
    4: "cryostate", 
    5: "cryo-vacuum",
    6: "Detector",
    7: "In Foil",
    8: "Cu plate",
    9: "Au Stage"

}

c = ROOT.TCanvas("c_all", "Energy Deposition for all Volumes", 1000, 700)
legend = ROOT.TLegend(0.70, 0.60, 0.95, 0.90) # Made it slightly larger for 9 entries
legend.SetBorderSize(1)

# ROOT has built-in palettes, or you can expand your list
colors = [ROOT.kRed, ROOT.kBlue, ROOT.kGreen+2, ROOT.kMagenta, ROOT.kCyan+2, 
          ROOT.kOrange+7, ROOT.kGray+2, ROOT.kYellow+2, ROOT.kBlack]

hist_list = []
first = True

# Loop from 1 to 9
for vlm in range(1, 10):
    hname = f"h_vlm_{vlm}"
    h = ROOT.TH1F(hname, "Energy Deposition;Energy (de);Counts", 100, 0, 700)
    
    t.Draw(f"de>>{hname}", f"vlm=={vlm}", "goff")
    
    entries = h.GetEntries()
    print(f"vlm {vlm}: {entries} entries")

    if entries == 0:
        continue

    # Set styles
    h.SetDirectory(0) 
    h.SetLineColor(colors[(vlm-1) % len(colors)])
    h.SetLineWidth(2)

    # Drawing logic
    if first:
        h.Draw("HIST")
        h.SetMaximum(t.GetEntries() * 0.05) # Adjust Y-axis scale if needed
        first = False
    else:
        h.Draw("HIST SAME")

    # Get name from dictionary, fallback to "vlm #" if not found
    display_name = vol_names.get(vlm, f"Volume {vlm}")
    legend.AddEntry(h, display_name, "l")
    hist_list.append(h)

legend.Draw()
c.Update()
c.Draw()
```
