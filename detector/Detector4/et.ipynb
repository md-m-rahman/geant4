```python
import ROOT
%jsroot on

f = ROOT.TFile.Open("tumor.root")
t = f.Get("t")

print("Tree entries =", t.GetEntries())
print("vlm min =", t.GetMinimum("vlm"))
print("vlm max =", t.GetMaximum("vlm"))

c = ROOT.TCanvas("c", "et for different volumes", 1000, 700)
legend = ROOT.TLegend(0.72, 0.68, 0.90, 0.90)
legend.SetBorderSize(1)
legend.SetFillStyle(0)

colors = [ROOT.kRed, ROOT.kBlue, ROOT.kGreen+2, ROOT.kMagenta, ROOT.kCyan+2, ROOT.kOrange+7]

hist_list = []
first = True

for i, vlm in enumerate([1,2,3]):   # change these if needed
    hname = f"h_vlm_{vlm}"
    h = ROOT.TH1F(hname, "et for different volumes;et;Counts", 100, 0, 700)

    t.Draw(f"et>>{hname}", f"vlm=={vlm}", "goff")
    print("vlm =", vlm, "entries =", h.GetEntries())

    if h.GetEntries() == 0:
        continue

    h.SetLineColor(colors[i % len(colors)])
    h.SetLineWidth(2)

    if first:
        h.Draw("hist")
        first = False
    else:
        h.Draw("hist same")

    legend.AddEntry(h, f"vlm = {vlm}", "l")
    hist_list.append(h)

legend.Draw()
c.Update()
c.Draw()
```
