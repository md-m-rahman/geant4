```python
import ROOT
%jsroot on
#=============================
# ROOT FILES
# ==========================================
file1 = "file1.root"
file2 = "file2.root"
file3 = "file3.root"

tree_name = "t"

# ==========================================
# OPEN FILES
# ==========================================
f1 = ROOT.TFile.Open("sucrose.root")
f2 = ROOT.TFile.Open("sucrose_urathane.root")
f3 = ROOT.TFile.Open("sucrose_cork.root")

t1 = f1.Get("t")
t2 = f2.Get("t")
t3 = f3.Get("t")
# ==========================================
# CANVAS
# ==========================================
c1 = ROOT.TCanvas("c_compare","Comparison of et[4]",1200,800)

# ==========================================
# HISTOGRAMS
# ==========================================
h1 = ROOT.TH1F(
    "h1",
    "et[4] comparison;et[4] (keV);Counts",
    200,
    0,
    700
)

h2 = ROOT.TH1F(
    "h2",
    "et[4] comparison;et[4] (keV);Counts",
    200,
    0,
    700
)

h3 = ROOT.TH1F(
    "h3",
    "et[4] comparison;et[4] (keV);Counts",
    200,
    0,
    700
)

# ==========================================
# COLORS
# ==========================================
h1.SetLineColor(ROOT.kBlue)
h2.SetLineColor(ROOT.kRed)
h3.SetLineColor(ROOT.kGreen+2)

h1.SetLineWidth(2)
h2.SetLineWidth(2)
h3.SetLineWidth(2)

# ==========================================
# DRAW HISTOGRAMS
# ==========================================

# File 1
t1.Draw(
    "et[4]>>h1",
    "pdg==22 && pro==2013"
)

# File 2
t2.Draw(
    "et[4]>>h2",
    "pdg==22 && pro==2013",
    "same"
)

# File 3
t3.Draw(
    "et[4]>>h3",
    "pdg==22 && pro==2013",
    "same"
)

# ==========================================
# LEGEND
# ==========================================
leg = ROOT.TLegend(0.65,0.72,0.90,0.90)
leg.AddEntry(h1, "polystyrene", "l")
leg.AddEntry(h2, "Urethane", "l")
leg.AddEntry(h3, "Cork", "l")
leg.Draw()

# ==========================================
# OPTIONAL
# ==========================================
ROOT.gStyle.SetOptStat(0)

leg.Draw()

c1.Modified()
c1.Update()
c1.Draw()
display(c1)
```
