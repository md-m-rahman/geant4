```python
import ROOT

# =========================
# OPEN ROOT FILE
# =========================
f = ROOT.TFile.Open("tumor.root")

if not f or f.IsZombie():
    raise RuntimeError("Could not open ROOT file")

t = f.Get("t")

if not t:
    raise RuntimeError("Could not find tree 't'")

# =========================
# CANVAS
# =========================
c1 = ROOT.TCanvas("c1",
                  "et for vlm==4",
                  1200,
                  800)

# =========================
# HISTOGRAM
# =========================
h1 = ROOT.TH1F(
    "h1",
    "et {vlm==4 && pdg==22 && pro==2013};et (keV);Counts",
    200,
    0,
    700
)

# =========================
# STYLE
# =========================
h1.SetLineColor(ROOT.kBlue)
h1.SetLineWidth(2)

ROOT.gStyle.SetOptStat(1111)

# =========================
# DRAW
# =========================
t.Draw(
    "et>>h1",
    "vlm==4 && pdg==22 && pro==2013"
)

# =========================
# UPDATE
# =========================
ROOT.gPad.Update()

# =========================
# SAVE IMAGE
# =========================
c1.SaveAs("vlm4_compton_gamma.png")

print("Histogram saved as vlm4_compton_gamma.png")

input("Press Enter to exit")
```
