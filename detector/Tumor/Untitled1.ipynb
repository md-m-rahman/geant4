```python
import ROOT
import numpy as np

# 1. Setup and File Loading
%jsroot on
f = ROOT.TFile.Open("tumor.root")
t = f.Get("t")

# 2. Canvas Setup
c_img = ROOT.TCanvas("c_img", "Tumor Reconstruction", 800, 800)
# We create a 2D Histogram to act as our "Image"
# 100 bins from -15cm to 15cm on both axes
image_map = ROOT.TH2F("image_map", "Back-Projection Reconstruction;X (cm);Y (cm)", 100, -15, 15, 100, -15, 15)
image_map.SetStats(0) # Hide stats for a cleaner image

# 3. Define the Energy Window (Photopeak)
# Only use gammas that deposited significant energy (ignores low-energy noise/scatter)
energy_cut = "et > 0.5" 

# 4. Perform Back-Projection
# For each detector, we draw a "beam" across the image area
print("Reconstructing tumor location...")

# Detector 4 (at X=+10, Y=0): Its field of view is a horizontal line across Y=0
t.Draw("0:iteration>>+image_map", f"vlm==4 && {energy_cut}", "goff")

# Detector 5 (at X=0, Y=+10): Its field of view is a vertical line across X=0
t.Draw("iteration:0>>+image_map", f"vlm==5 && {energy_cut}", "goff")

# Detector 6 (at X=-10, Y=0): Similar to Det 4
t.Draw("0:iteration>>+image_map", f"vlm==6 && {energy_cut}", "goff")

# Detector 7 (at X=0, Y=-10): Similar to Det 5
t.Draw("iteration:0>>+image_map", f"vlm==7 && {energy_cut}", "goff")

# 5. Visualization
# "COLZ" draws it as a heat map with a color bar
image_map.Draw("COLZ")

# Optional: Smooth the image to make the tumor look more realistic
image_map.Smooth(1)

c_img.Draw()

print("Reconstruction complete. The 'hot spot' at the center indicates the tumor.")
```
