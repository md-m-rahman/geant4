```python
import ROOT
from collections import Counter
from collections import Counter

def check_volumes(root_file):
    f = ROOT.TFile.Open(root_file)
    t = f.Get("t")
    c = Counter()

    for entry in t:
        n = min(len(entry.vlm), len(entry.pdg))
        for i in range(n):
            c[int(entry.vlm[i])] += 1

    print(root_file, c)
    f.Close()

check_volumes("tumor.root")
check_volumes("phantom.root")
```
