```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from scipy.special import spherical_jn
from scipy.integrate import cumulative_trapezoid

# =====================================================
# 1. File path
# =====================================================

csv_file = "/root/geant4/fac/Nb_all_wavefunctions.csv"
# If it is CSV, use:
# filename = r"E:\geant4\fac\Kr\Kr_all_wavefunctions.csv"

# =====================================================
# 2. Read file
# =====================================================

if filename.endswith(".xlsx"):
    df = pd.read_excel(filename, header=None)
elif filename.endswith(".csv"):
    df = pd.read_csv(filename, header=None)
else:
    raise ValueError("File must be .xlsx or .csv")

df.columns = ["orbital", "index", "r", "Vr", "Veffr", "P", "Q"]

print("Available orbitals:")
print(df["orbital"].unique())

# =====================================================
# 3. Choose orbital
# =====================================================

orbital_name = "Kr_2p12"   # change this if needed

d = df[df["orbital"] == orbital_name].copy()

if len(d) == 0:
    raise ValueError("Orbital name not found. Check printed orbital list.")

r = d["r"].to_numpy(dtype=float)
P = d["P"].to_numpy(dtype=float)   # large component = rG
Q = d["Q"].to_numpy(dtype=float)   # small component = rF

# Sort by r
idx = np.argsort(r)
r = r[idx]
P = P[idx]
Q = Q[idx]

# =====================================================
# 4. Choose angular momentum
# =====================================================

# For Kr_2p12:
l_large = 1   # p orbital
l_small = 0   # p1/2 small component

# For Kr_2p32 use:
# l_large = 1
# l_small = 2

# =====================================================
# 5. Check normalization
# =====================================================

norm = np.trapezoid(P**2 + Q**2, r)

print("\nNormalization:")
print("∫(P² + Q²) dr =", norm)

# =====================================================
# 6. Momentum grid
# =====================================================

p_grid = np.arange(0.0, 100.0 + 0.05, 0.05)

const = np.sqrt(2.0 / np.pi)

chiG = []
chiF = []

# =====================================================
# 7. Calculate ChiG and ChiF individually
# =====================================================

for p in p_grid:

    x = p * r

    j_large = spherical_jn(l_large, x)
    j_small = spherical_jn(l_small, x)

    integrand_G = P * j_large * r
    integrand_F = Q * j_small * r

    chi_g = const * np.trapezoid(integrand_G, r)
    chi_f = const * np.trapezoid(integrand_F, r)

    chiG.append(chi_g)
    chiF.append(chi_f)

chiG = np.array(chiG)
chiF = np.array(chiF)

# =====================================================
# 8. Total Chi
# =====================================================

chiTotal = np.sqrt(chiG**2 + chiF**2)

# =====================================================
# 9. I(p)
# =====================================================

I_p = chiTotal**2 * p_grid**2

# =====================================================
# 10. J(Q)
# =====================================================

J_integrand = chiTotal**2 * p_grid

J_Q = -0.5 * cumulative_trapezoid(
    J_integrand[::-1],
    p_grid[::-1],
    initial=0
)[::-1]

# =====================================================
# 11. Print values
# =====================================================

print("\n p        ChiF              ChiG              ChiTotal          I(p)              J(Q)")
print("-----------------------------------------------------------------------------------------")

for i in range(0, len(p_grid), 20):
    print(
        f"{p_grid[i]:6.2f}   "
        f"{chiF[i]: .8E}   "
        f"{chiG[i]: .8E}   "
        f"{chiTotal[i]: .8E}   "
        f"{I_p[i]: .8E}   "
        f"{J_Q[i]: .8E}"
    )

# =====================================================
# 12. Save result
# =====================================================

out = pd.DataFrame({
    "p_or_Q": p_grid,
    "ChiF": chiF,
    "ChiG": chiG,
    "ChiTotal": chiTotal,
    "I(p)": I_p,
    "J(Q)": J_Q
})

output_name = orbital_name + "_step_by_step_result.csv"
out.to_csv(output_name, index=False)

print("\nSaved:", output_name)

# =====================================================
# 13. Plot
# =====================================================

plt.figure(figsize=(7,5))
plt.plot(p_grid, chiF, label="ChiF small")
plt.plot(p_grid, chiG, label="ChiG large")
plt.plot(p_grid, chiTotal, label="Chi total")
plt.xlabel("p (a.u.)")
plt.ylabel("Chi")
plt.xlim(0, 20)
plt.legend()
plt.grid()
plt.show()

plt.figure(figsize=(7,5))
plt.plot(p_grid, I_p)
plt.xlabel("p (a.u.)")
plt.ylabel("I(p)")
plt.xlim(0, 20)
plt.grid()
plt.show()

plt.figure(figsize=(7,5))
plt.plot(p_grid, J_Q)
plt.xlabel("Q (a.u.)")
plt.ylabel("J(Q)")
plt.xlim(0, 20)
plt.grid()
plt.show()
```
