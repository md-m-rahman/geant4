```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from scipy.special import spherical_jn
from scipy.integrate import cumulative_trapezoid

# =====================================================
# Read FAC relativistic Kr wavefunctions
# =====================================================

csv_file = "/root/geant4/fac/Kr_all_wavefunctions.csv"
df = pd.read_csv(csv_file)

df = df.dropna(subset=["r", "P", "Q"])

# =====================================================
# Momentum and Q grid
# =====================================================

p = np.arange(0.0, 1000.0 + 0.05, 0.05)
Qgrid = np.arange(0.0, 100.0 + 0.05, 0.05)

# =====================================================
# Functions
# =====================================================

def get_orbital(df, name):
    d = df[df["Orbital"] == name].copy()
    d = d.sort_values("r")

    r = d["r"].to_numpy()
    G = d["P"].to_numpy()
    F = d["Q"].to_numpy()

    norm = np.trapezoid(r**2 * (G**2 + F**2), r)
    G = G / np.sqrt(norm)
    F = F / np.sqrt(norm)

    print(name, "norm before =", norm)

    return r, G, F


def chi_large(p_grid, r, G, l):
    out = []
    for pp in p_grid:
        jl = spherical_jn(l, pp*r)
        val = np.sqrt(2/np.pi) * np.trapezoid(G * jl * r**2, r)
        out.append(val)
    return np.array(out)


def chi_small(p_grid, r, F, l, jtype):
    if jtype == "plus":
        lp = l + 1
    elif jtype == "minus":
        lp = l - 1
    else:
        raise ValueError("jtype must be plus or minus")

    out = []
    for pp in p_grid:
        jl = spherical_jn(lp, pp*r)
        val = np.sqrt(2/np.pi) * np.trapezoid(F * jl * r**2, r)
        out.append(val)
    return np.array(out)


def calculate_J_rel(p_grid, chiG, chiF, Qgrid):
    integrand = (chiG**2 + chiF**2) * p_grid

    rev = cumulative_trapezoid(
        integrand[::-1],
        p_grid[::-1],
        initial=0
    )

    J_p = -0.5 * rev[::-1]
    J_Q = np.interp(Qgrid, p_grid, J_p)

    return J_Q


def one_orbital_J(df, name, l, jtype):
    r, G, F = get_orbital(df, name)

    chiG = chi_large(p, r, G, l)
    chiF = chi_small(p, r, F, l, jtype)

    J = calculate_J_rel(p, chiG, chiF, Qgrid)

    check = 2*np.trapezoid(J, Qgrid)
    print(name, "2∫JdQ =", check)

    return J


# =====================================================
# Calculate Kr relativistic orbitals
# Check your actual names using print(df["Orbital"].unique())
# =====================================================

print(df["Orbital"].unique())

J_1s12 = one_orbital_J(df, "Kr_1s12", l=0, jtype="plus")

J_2s12 = one_orbital_J(df, "Kr_2s12", l=0, jtype="plus")
J_2p12 = one_orbital_J(df, "Kr_2p12", l=1, jtype="minus")
J_2p32 = one_orbital_J(df, "Kr_2p32", l=1, jtype="plus")

J_3s12 = one_orbital_J(df, "Kr_3s12", l=0, jtype="plus")
J_3p12 = one_orbital_J(df, "Kr_3p12", l=1, jtype="minus")
J_3p32 = one_orbital_J(df, "Kr_3p32", l=1, jtype="plus")
J_3d32 = one_orbital_J(df, "Kr_3d32", l=2, jtype="minus")
J_3d52 = one_orbital_J(df, "Kr_3d52", l=2, jtype="plus")

J_4s12 = one_orbital_J(df, "Kr_4s12", l=0, jtype="plus")
J_4p12 = one_orbital_J(df, "Kr_4p12", l=1, jtype="minus")
J_4p32 = one_orbital_J(df, "Kr_4p32", l=1, jtype="plus")


# =====================================================
# Relativistic occupation for Kr
# Kr = 36 electrons
# =====================================================

J_total_Kr_rel = (
    2*J_1s12
    + 2*J_2s12
    + 2*J_2p12 + 4*J_2p32
    + 2*J_3s12
    + 2*J_3p12 + 4*J_3p32
    + 4*J_3d32 + 6*J_3d52
    + 2*J_4s12
    + 2*J_4p12 + 4*J_4p32
)

print("Total Kr check 2∫Jtotal dQ =", 2*np.trapezoid(J_total_Kr_rel, Qgrid))
print("Expected = 36")


# =====================================================
# Save result
# =====================================================

table = pd.DataFrame({
    "Q": Qgrid,
    "J_1s1/2(2)": 2*J_1s12,
    "J_2s1/2(2)": 2*J_2s12,
    "J_2p1/2(2)": 2*J_2p12,
    "J_2p3/2(4)": 4*J_2p32,
    "J_3s1/2(2)": 2*J_3s12,
    "J_3p1/2(2)": 2*J_3p12,
    "J_3p3/2(4)": 4*J_3p32,
    "J_3d3/2(4)": 4*J_3d32,
    "J_3d5/2(6)": 6*J_3d52,
    "J_4s1/2(2)": 2*J_4s12,
    "J_4p1/2(2)": 2*J_4p12,
    "J_4p3/2(4)": 4*J_4p32,
    "J_total_Kr_relativistic": J_total_Kr_rel
})

table.to_csv("Krypton_relativistic_Compton_profile.csv", index=False)
print(table)


# =====================================================
# Plot
# =====================================================

plt.figure(figsize=(8,6))

plt.plot(Qgrid, 2*J_1s12, label="1s1/2(2)")
plt.plot(Qgrid, 2*J_2s12, label="2s1/2(2)")
plt.plot(Qgrid, 2*J_2p12 + 4*J_2p32, label="2p(6)")
plt.plot(Qgrid, 2*J_3s12, label="3s1/2(2)")
plt.plot(Qgrid, 2*J_3p12 + 4*J_3p32, label="3p(6)")
plt.plot(Qgrid, 4*J_3d32 + 6*J_3d52, label="3d(10)")
plt.plot(Qgrid, 2*J_4s12, label="4s1/2(2)")
plt.plot(Qgrid, 2*J_4p12 + 4*J_4p32, label="4p(6)")
plt.plot(Qgrid, J_total_Kr_rel, label="Kr total relativistic", linewidth=2)

plt.xlabel("Q (a.u.)")
plt.ylabel("J(Q)")
plt.title("Relativistic Krypton Compton Profile from FAC G and F")
plt.yscale("log")
plt.grid(True)
plt.legend()
plt.show()
```
